When we need to store several values in the same variable, we can use _lists_. A list is an ordered collection of values. It might contain, for example, element names, temperatures from an experiment, or absorbance values measured at different wavelengths.

Lists in Python are enclosed in square brackets (`[]`), with elements separated by commas:

`````python
elements = ["hydrogen", "helium", "lithium", "beryllium", "boron"]
atomic_numbers = [1, 2, 3, 4, 5]
temperatures = [20.0, 25.0, 30.0, 35.0]
`````

The order of the elements is preserved. In the list `elements`, "hydrogen" is the first element and "boron" is the last. A list can contain numbers, text, and other data types, but when a list represents measured data, it is usually clearest if all values have the same meaning and unit.

```{admonition} List
A list is an ordered collection of values stored in one variable. Lists can be changed after they are created.
```

## Adding and removing elements

Lists support operations such as adding, removing, and rearranging elements. Here we use a list of chemical elements as an example:

`````python
elements = []                                  # Create an empty list
noble_gases = ["helium", "neon", "argon"]   # Create a list of noble gases

elements.append("lithium")                   # Add one element
elements.extend(noble_gases)                   # Add all elements from noble_gases
elements.remove("neon")                      # Remove the element with value "neon"
removed_element = elements.pop(1)              # Remove the element at index 1
elements.reverse()                             # Reverse the order

print(elements)
print("Removed element:", removed_element)
`````

```{admonition} Exercise
:class: tip
What output does the program above produce? Try to follow the changes to the list without using a computer. Then run the program and check your prediction.
```

````{admonition} Suggested solution
:class: tip, dropdown
After `append` and `extend`, the list is `["lithium", "helium", "neon", "argon"]`. First, "neon" is removed by value. Then `pop(1)` removes the element at index 1, which is "helium", and stores it in `removed_element`. Finally, the list is reversed. The program therefore prints:

```{code-block} text
['argon', 'lithium']
Removed element: helium
```
````

Notice the difference between `remove` and `pop`: `remove` uses the value itself, while `pop` uses an index. If the same value occurs more than once, `remove` removes the first occurrence.

## Indices

The position number of an element in a list is called an _index_. We specify an index in square brackets. We can also find the length of a list and search for the index of a particular value:

`````python
elements = ["H", "C", "N", "O", "F", "Na", "Mg", "Cl", "Ar"]
atomic_numbers = [1, 6, 7, 8, 9, 11, 12, 17, 18]

print("Number of elements:", len(elements))
print("Index of Cl:", elements.index("Cl"))
print("Atomic number of Cl:", atomic_numbers[7])
`````

The list contains nine elements, and chlorine is in position 8 if we count in the ordinary way. Its index is nevertheless 7 because indexing in Python, as in many programming languages, starts at 0. The first element has index 0 and the final element in this list has index 8. A list containing $n$ elements therefore has indices from $0$ to $n-1$.

The two lists above are **parallel lists**: elements with the same index belong together. The symbol `"Cl"` has index 7 in `elements`, so the atomic number of chlorine is stored at index 7 in `atomic_numbers`. The same idea can be used to pair, for example, wavelengths and absorbance values.

## Retrieving parts of a list

We do not need to process an entire list every time we want to use some of its elements. A single index retrieves one element, while a colon can be used to retrieve an interval, called a _slice_:

`````python
absorbances = [0.12, 0.25, 0.37, 0.49, 0.61]

A = absorbances[2]      # Element at index 2
B = absorbances[1:3]    # From index 1 up to, but not including, index 3
C = absorbances[2:]     # From index 2 to the end
D = absorbances[:2]     # From the beginning up to, but not including, index 2
E = absorbances[-1]     # Final element
F = absorbances[-2]     # Second-to-last element
`````

```{admonition} Exercise
:class: tip
What values do A, B, C, D, E, and F have?
```

````{admonition} Suggested solution
:class: tip, dropdown
Indices specify positions in a list. A negative index counts backwards from the end. In a slice, the stop index is excluded. Thus `[1:3]` contains the elements at indices 1 and 2, but not the element at index 3.

If nothing is written before the colon, the slice starts at the first element. If nothing is written after the colon, it continues to the end. Therefore:

```{code-block} text
A: 0.37
B: [0.25, 0.37]
C: [0.37, 0.49, 0.61]
D: [0.12, 0.25]
E: 0.61
F: 0.49
```
````

## Combining and inspecting lists

There are often several ways to do the same thing. Consider an example in which we combine two temperature series and then modify the new list:

`````python
temperature_series1 = [20.0, 25.0, 30.0]
temperature_series2 = [35.0, 40.0]

temperatures = temperature_series1 + temperature_series2
temperatures.append(45.0)
temperatures.remove(30.0)

print(temperatures)
print("Is 30.0 in the list?", 30.0 in temperatures)
`````

```{admonition} Exercise
:class: tip
Explain what happens in the program above. What is printed? Extend the program so that it also prints the final temperature and all temperatures except the first.
```

````{admonition} Suggested solution
:class: tip, dropdown
The `+` operator combines the two lists. Then 45.0 is added and 30.0 is removed. The program prints `[20.0, 25.0, 35.0, 40.0, 45.0]`, followed by `False`, because 30.0 is no longer in the list.

The two additional print statements can be written as:

```{code-block} Python
print(temperatures[-1])
print(temperatures[1:])
```
````

When we add lists using `+`, we concatenate them. We do not add corresponding numerical elements:

`````python
[1, 2] + [3, 4]  # Gives [1, 2, 3, 4], not [4, 6]
`````

Later we will use NumPy arrays when we want to perform the same mathematical operation on many measured values at once.

### Some useful list operations

The table below shows a few more list operations. You do not need to memorise them. The important skill is to read code, recognise the kind of operation you need, and look up the exact command when necessary.

| Operation | Meaning | Example |
|---|---|---|
| `len(list)` | Number of elements | `len(temperatures)` |
| `sum(list)` | Sum of all numerical values | `sum(titration_volumes)` |
| `min(list)` | Smallest value | `min(pH_values)` |
| `max(list)` | Largest value | `max(absorbances)` |
| `value in list` | Tests whether a value is present | `"Cl" in elements` |
| `list.index(value)` | Index of the first occurrence | `elements.index("Cl")` |
| `list.count(value)` | Number of occurrences | `atoms.count("H")` |
| `list.sort()` | Sorts and modifies the original list | `temperatures.sort()` |

```{admonition} Lists can be modified
:class: caution
If we write `copy = measurements`, we do not create an independent copy. Both variable names refer to the same list, so a change through one name is visible through the other. Use `copy = measurements.copy()` if you need a separate list that can be changed independently.
```
