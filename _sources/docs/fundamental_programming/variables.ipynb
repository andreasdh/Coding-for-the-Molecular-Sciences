## Variables

One of the most fundamental ideas in programming is the _variable_. We will use variables in almost every program we write. A variable can store a value, for example a number, a piece of text, or a list containing many numbers.

When we write `temperature = 298`, we create a variable called `temperature` and assign it the value 298. Here the equals sign is an **assignment operator**: the value on the right is assigned to the variable on the left. This means that `temperature = 298` is not a mathematical equation, but an instruction to the computer.

```{admonition} Variable
A programming variable is a named quantity that stores a value in a program.
```

Variable names can contain letters, numbers, and underscores, but they cannot begin with a number or contain spaces. Python is also case-sensitive, so `temperature` and `Temperature` are two different variables. Use names that communicate what a value represents. For example, `HCl_concentration` is more informative than simply `x`.

Variables are particularly useful when we need the same value several times or when we update a value while the program is running. In a chemistry program, a variable might represent a mass, temperature, concentration, or volume. If we add titrant in 1 mL increments, we might describe the process using pseudocode like this:

```{code-block} text
volume_added = 0
repeat 10 times:
    volume_added = volume_added + 1
    display volume_added on the screen
```

The value stored in `volume_added` is updated after each addition. The instruction `volume_added = volume_added + 1` would make no sense as a mathematical equation, but in a program it means: retrieve the old value, increase it by 1, and store the result in the same variable. You can think of a variable as a labelled box whose contents can be retrieved and changed while the program runs.

```{admonition} What is pseudocode?
:class: caution, dropdown
Pseudocode is a description of an algorithm that does not use a particular programming language. It can be written in many ways and may use words, diagrams, and symbols.
```

## Comments
Comments are important for remembering what code illustrates, which units are being used, and what assumptions have been made. They are also valuable when other people need to understand or reuse your code. In chemical calculations, commenting units and assumptions is particularly useful because Python does not keep track of physical units for us.

A comment begins with `#`. If a comment spans several lines, each line should normally begin with `#`. Triple quotation marks can be used to create a multi-line string. Such a string at the beginning of a program or function can serve as documentation describing what the code does. Here is an example:

`````python
"""
Calculate the heat required to increase the temperature
of a water sample, neglecting heat loss.
"""

mass = 100.0              # mass of water in g
heat_capacity = 4.18      # specific heat capacity in J/(g*K)
temperature_change = 10   # temperature change in K

heat = mass * heat_capacity * temperature_change
print("The water sample absorbs", heat, "J.")
`````

## Arithmetic
We can use Python as a simple calculator. Some of the most important operators are:

| Operator | Meaning |
|---|---|
| + | Addition (`a + b`) |
| - | Subtraction (`a - b`) |
| * | Multiplication (`a*b`) |
| / | Division (`a/b`) |
| ** | Exponentiation (`a**b`) |

Python follows the usual order of operations: powers are evaluated before multiplication and division, which are evaluated before addition and subtraction. Use parentheses when you want to change the order or make an expression easier to read. For example, `n*R*T/V` and `(n*R*T)/V` give the same result, but the parentheses may make the relationship to a chemical formula clearer.

In program code we use a decimal point. The number 0.25 is therefore written as `0.25`.

`//` performs floor division. For positive integers, it gives the number of complete times the divisor fits into the dividend. Thus `24 // 5 = 4`, because $4\times5=20$. The remainder is 4. **Example from the chemistry laboratory**

Suppose we have 24 mL of solution and want to fill sample vials with 5 mL in each. Because the volumes in this simplified example are whole millilitres, we can use floor division to find the number of full vials and modulus to find the remaining volume:

`````python
total_volume = 24       # mL
volume_per_vial = 5     # mL

number_of_full_vials = total_volume // volume_per_vial
remaining_volume = total_volume % volume_per_vial

print("We can fill", number_of_full_vials, "sample vials.")
print("There will be", remaining_volume, "mL of solution left.")
`````

The program shows that we can fill four sample vials and that 4 mL remains. In the laboratory, volumes are often measured as decimal values, in which case ordinary division (`/`) is usually more relevant. Floor division and modulus are mainly useful when distributing a whole number of units or determining how much remains.

### Libraries
Many mathematical functions are not built directly into the Python language. To access functions such as `sqrt`, `cos`, and `log`, we import _libraries_ or _modules_. Python has a standard library module called `math`, but in this material we often use NumPy because it can also perform calculations on many numbers at once. This becomes particularly useful when we later work with data sets and simulations.

There are several ways to import a library. Here are some ways to import the very useful _NumPy_ library (**num**erical **Py**thon):

`````python
# Method 1: Import only the functions you need
from numpy import log10

H3O_concentration = 1E-7  # mol/L
pH = -log10(H3O_concentration)

# Method 2: Import the entire library
import numpy

R = 8.314       # J/(mol*K)
T = 298         # K
M_O2 = 0.032    # kg/mol
v_rms = numpy.sqrt(3*R*T/M_O2)

# Method 3: Import the library using an alias.
# This is the most common approach.
import numpy as np

A = 1E12        # pre-exponential factor in 1/s
E_a = 75000     # activation energy in J/mol
k = A*np.exp(-E_a/(R*T))

# Method 4: Import everything without specifying
# which library the functions come from
from numpy import *

pH = -log10(1E-7)
v_rms = sqrt(3*R*T/M_O2)
`````

Method 3 is the most common approach and the one we will use most often. Some older teaching videos use a star import from a collection called _pylab_ (`from pylab import *`). Consider the root-mean-square speed of gas molecules, $v_{rms}$. It can be calculated from:

$$v_{rms}=\sqrt{\frac{3RT}{M}}$$

where $R=8.314$ J mol$^{-1}$ K$^{-1}$, $T$ is the temperature in K, and $M$ is the molar mass in kg/mol.

## User input
We can ask the user of a program to enter values. This can be useful, for example, when the same program will be used to calculate the amount of substance for several samples with different masses. We obtain user input with the `input` function.

Start by assigning fixed values to the variables and test the program. Then replace selected values with input. This avoids having to re-enter values every time you run code while debugging it.

### F-string formatting

When we combine text and variables in printed output, we can separate them with commas as above. Another, often clearer method is a **formatted string**, usually called an _f-string_. Put the letter `f` before the opening quotation mark and write variables or expressions inside braces:

`````python
print(f"{mass} g {substance} corresponds to {amount} mol.")
`````

Python replaces the contents of the braces with the values stored in the variables. This lets us write the complete sentence as one continuous string. We can also control how many decimal places are displayed.

`````python
substance = "NaCl"
mass = 2.922       # g
molar_mass = 58.44 # g/mol

amount = mass / molar_mass

print(f"{mass} g {substance} corresponds to {amount} mol.")
print(f"{mass:.2f} g {substance} corresponds to {amount:.3f} mol.")
`````

In the second print statement, `:.2f` means that the value is displayed as a floating-point number with two decimal places, while `:.3f` gives three decimal places. Formatting changes only how the number is displayed, not the value stored in the variable. It is usually best to retain full precision during calculations and round only when presenting the result.

## Data types
In mathematics we work with number sets such as real, rational, integer, natural, irrational, and complex numbers. Many of these sets contain infinitely many numbers, while a computer has finite storage. Python therefore uses different data types to represent numbers and other values. A floating-point number (`float`) is, for example, an approximate representation of a real number, so some calculations can produce small rounding errors. The most important data types here are:

| Data type | Meaning | Example |
|---|---|---|
| Integer (`int`) | Whole number | `number_of_C = 12` |
| Floating-point number (`float`) | Decimal number | `concentration = 0.10` |
| String (`str`) | Text | `substance = "NaCl"` |
| Boolean (`bool`) | True or false | `is_acidic = True` |

When we obtain input from a user, we need to consider which data type we want. The default is text. If we "add" the text strings `"4"` and `"2"`, we obtain `"42"`, whereas the number 4 plus the number 2 gives 6. The example below illustrates the difference.

`````python
text1 = input("Enter a number: ")
text2 = input("Enter another number: ")

number1 = float(text1)
number2 = float(text2)

text_result = text1 + text2
numeric_result = number1 + number2

print("The sum of the text strings is:", text_result,
      "and the sum of the numbers is:", numeric_result)
`````

## Computers require precise instructions
When you create programs in Python, remember the following:
- The program is read from top to bottom and from left to right.
- Lower-case and upper-case letters are different (`p` $\neq$ `P`).
- Python usually ignores extra spaces, except at the beginning of a line where indentation matters. Use spaces to make code readable.
- Python does not know whether a number represents grams, litres, or mol/L. Use clear variable names and comments to keep track of units.
- Practise reading error messages: they usually identify the line involved and the type of error.
