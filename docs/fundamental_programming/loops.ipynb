## Definition

Loops are among the most important structures in programming. They allow a computer to repeat an operation as many times as needed, which is one of the computer's great strengths.

Repetition occurs throughout chemistry and molecular science. We may perform the same calculation for every wavelength in a spectrum, inspect every measurement in a data set, update concentrations over many small time steps, or simulate the motion of many particles. Without loops, we would have to write the same code again for every measurement, time step, or particle.

```{admonition} Loops
A loop is a structure that repeats code. A *counting loop* repeats something a specified number of times, while a *condition-controlled loop* repeats something as long as a condition is true. In Python, these are normally written as `for` and `while` loops, respectively.
```

We begin with a visual representation of repetition before applying the same principles to chemical data and models.

## Turtle graphics

We will first explore loops using a turtle. It follows simple commands such as `forward`, `backward`, `right`, and `left`.

`````python
from turtle import *

shape("turtle")     # Give the pointer a turtle shape
color("limegreen")  # Make the turtle lime green
forward(80)            # Move forward by 80 pixels
left(60)               # Turn 60 degrees to the left
`````

The first line imports commands from the `turtle` library. A library is a collection of pre-written functions that we can use in our programs. We did not create `forward` or `left`; they are defined in `turtle`.

### A simplified representation of a benzene ring

A useful chemical starting point is the hexagonal geometry associated with benzene. Without a loop, we would have to repeat `forward(80)` and `left(60)` six times. With a loop, we can write:

`````python
from turtle import *

shape("turtle")
color("limegreen")
speed(3)

number_of_sides = 6
side_length = 80
turning_angle = 360 / number_of_sides

for side in range(number_of_sides):
    forward(side_length)
    left(turning_angle)

done()
`````

The loop repeats the indented statements six times. On the first iteration, `side = 0`; then `side` takes the values 1, 2, 3, 4, and 5. Thus `range(6)` produces six integers from 0 up to, but not including, 6.

## Randomness and the `random` library

The motion of particles in a liquid or gas is influenced by an enormous number of collisions. A simple model cannot follow all of them directly. Instead, we can allow a model particle to choose a random direction at each step. Python's standard `random` library provides tools for this. We begin with `choice`:

`````python
from random import choice

possible_directions = [0, 90, 180, 270]
direction = choice(possible_directions)

print(direction)
`````

`choice` selects one random element from a list, so repeated runs may produce different directions. Computers generate *pseudorandom* values using algorithms, but they behave randomly enough for this simple model. In more advanced work it can be useful to control the random seed so that a simulation can be reproduced exactly.

### A random walk as a model of diffusion

We can combine `choice`, `turtle`, and a `for` loop:

`````python
from turtle import *
from random import choice

shape("turtle")
speed(0)
pensize(2)

possible_directions = [0, 90, 180, 270]
number_of_steps = 200
step_length = 10

dot(10, "red")             # Starting point

for step in range(number_of_steps):
    direction = choice(possible_directions)
    setheading(direction)
    forward(step_length)

dot(10, "blue")            # Final point
done()
`````

At each step, the program selects one of four directions with equal probability. `setheading` points the turtle in the selected direction and `forward` moves it one step. The red dot marks the start and the blue dot the finish.

This is a *random walk*. It can illustrate how many random motions produce a net displacement over time, but the model is highly simplified:

- a molecule moves in three dimensions, while the turtle moves in two;
- molecules can move in far more than four directions;
- molecular motion does not consist of equal-length steps on a grid;
- one trajectory describes one model particle, not the spreading of an ensemble.

The model is therefore not a literal picture of molecular motion, but it can still help us explore basic features of diffusion.

## `while` loops

A `while` loop is condition-controlled: it repeats code while a condition remains true. It is useful when we do not know in advance exactly how many repetitions will be needed.

Suppose we perform a series of 1:2 dilutions. After each dilution, the concentration is half its previous value. We want to find how many dilutions are needed before the concentration falls below a threshold.

`````python
concentration = 0.80       # mol/L
limit = 0.05               # mol/L
number_of_dilutions = 0

while concentration >= limit:
    print(f"After {number_of_dilutions} dilutions: {concentration:.3f} mol/L")
    concentration = concentration / 2
    number_of_dilutions = number_of_dilutions + 1

print(f"The concentration is now {concentration:.3f} mol/L.")
`````

Before every iteration, Python checks whether `concentration >= limit` is true. If it is true, the indented statements run and the condition is checked again. When the concentration falls below the limit, the loop stops.

Something inside a `while` loop must be able to make the condition false. If we forget to update `concentration`, the condition remains true and we create an *infinite loop*.

## `for` loops

A `for` loop is a counting loop. It is useful when we know how many times something should be repeated or when we want to perform the same operation for every item in a collection.

### The `range` function

`range` generates a sequence of integers that a loop can traverse:

| Command | Values |
|---|---|
| `range(3)` | 0, 1, 2 |
| `range(2, 4)` | 2, 3 |
| `range(1, 8, 2)` | 1, 3, 5, 7 |

A general form is `range(start, stop, step)`. The stop value is excluded. With one argument, the sequence begins at 0. If no step is given, the step is 1.

`````python
for measurement in range(5):
    print(f"Performing measurement number {measurement + 1}")
`````

The variable `measurement` takes the values 0, 1, 2, 3, and 4. We add 1 in the printed output because people normally call the first measurement measurement 1 rather than measurement 0.

### Looping through lists

We do not always need `range`. To process every value in a list, the loop variable can take the values directly:

`````python
absorbance_values = [0.42, 0.39, 0.44, 0.41]
total_absorbance = 0

for absorbance in absorbance_values:
    total_absorbance = total_absorbance + absorbance

mean_absorbance = total_absorbance / len(absorbance_values)
print(f"Mean absorbance: {mean_absorbance:.3f}")
`````

On the first iteration, `absorbance = 0.42`; on the next, it is 0.39, and so on until every element has been used. `total_absorbance` is updated after each measurement.

## Thinking in loops

Loops can sum sequences, process measurements, and explore how chemical systems develop over time. To use a loop, we divide a problem into a small set of steps that can be repeated. In mathematical science we often represent relationships using continuous functions and equations; in code, we can instead perform a small operation at a time. This is an *iterative* way of solving a problem.

### Processing chemical data

Suppose absorbance values from 0.10 to 1.00 lie within the recommended working range for a particular method. We can select them as follows:

`````python
absorbance_values = [0.08, 0.42, 0.39, 1.18, 0.44]
valid_measurements = []

for absorbance in absorbance_values:
    if 0.10 <= absorbance <= 1.00:
        valid_measurements.append(absorbance)

print(valid_measurements)
`````

The list, loop, and conditional work together: each absorbance is checked in turn, and only values satisfying the criterion are added to the new list. The range is merely an example; in a real analysis it must be determined from the instrument, calibration, and method. A program can apply a criterion consistently, but it cannot decide whether the criterion is scientifically appropriate.

### A simple kinetic model

We can also use a loop to explore how a reaction changes with time. Consider the irreversible reaction

$$A \rightarrow B$$

with rate $v=k[A]$. During a small time step $\Delta t$, we approximate

$$\Delta[A] \approx -k[A]\Delta t.$$

We can therefore update the concentrations step by step:

`````python
k = 0.15             # rate constant in s^-1
A = 1.00             # initial concentration of A in mol/L
B = 0.00             # initial concentration of B in mol/L
dt = 0.10            # time step in s
end_time = 10        # final time in s

number_of_steps = int(end_time / dt)

for step in range(number_of_steps):
    change = k * A * dt
    A = A - change
    B = B + change

print(f"After {end_time} s, [A] = {A:.3f} mol/L.")
print(f"After {end_time} s, [B] = {B:.3f} mol/L.")
print(f"Total concentration: {A + B:.3f} mol/L.")
`````

At every time step, we calculate how much A is converted. The same change is subtracted from A and added to B, conserving total concentration in this simplified model. This is a numerical approximation: we assume, among other things, that the rate is approximately constant within each small time step. If `dt` is too large, the approximation becomes poorer.

```{admonition} Exercise
:class: tip

```

```{admonition} Suggested solution
:class: tip, dropdown
As the time step becomes smaller, the rate is updated more frequently and the result approaches a stable value. A smaller time step requires more iterations. The final time must remain the same, so `number_of_steps = end_time / dt`. With an end time of 10 s, `dt = 1` gives ten steps while `dt = 0.01` gives one thousand.
```

## Sequences and series

Loops are often used to generate sequences. After each half-life, for example, the fraction of a radioactive substance remaining is

$$1,\frac12,\frac14,\frac18,\frac1{16},\ldots$$

If $N_n$ is the amount after $n$ half-lives, the sequence can be written recursively as $N_{n+1}=N_n/2$. A recursive relationship uses an earlier value to calculate the next one, which maps naturally onto a loop:

`````python
amount = 4.0  # kg

for number_of_half_lives in range(11):
    print(f"n = {number_of_half_lives}: {amount:.6f} kg")
    amount = amount / 2
`````

A sequence whose terms are added is a *series*. Consider the geometric series

$$1+\frac23+\frac49+\frac8{27}+\ldots$$

whose $n$th term is $(2/3)^n$. A computer cannot sum infinitely many terms, but we can approximate the sum using a large finite number:

`````python
number_of_terms = 100
total = 0

for n in range(number_of_terms):
    term = (2 / 3)**n
    total = total + term

print(total)
`````

The analytical sum is 3. The program produces a value extremely close to 3, but it is still a numerical approximation. The example illustrates a general pattern: start with an initial value, calculate a new contribution, and update an accumulated value on every iteration.

## Nested loops

A nested loop is a loop inside another loop. The inner loop is repeated for every iteration of the outer loop.

For example, an electron beginning in hydrogen energy level $n$ may de-excite to any lower level $m<n$. We can print all downward transitions from levels 2 through 5:

`````python
for n in range(2, 6):
    for m in range(1, n):
        print(f"Possible transition: n = {n} to m = {m}")
`````

For `n = 2`, the inner loop contains only `m = 1`; for `n = 3`, it contains 1 and 2. The inner loop therefore becomes longer for each new energy level. The program prints $1+2+3+4=10$ transitions.

````{admonition} Suggested solution
```{code-block} Python

```

Do not use a nested loop when a simpler structure solves the problem.
````

````{admonition} Suggested solution
plt.plot(pressures_kPa, volumes_L, marker="o")
plt.xlabel("Pressure (kPa)")
plt.ylabel("Volume (L)")
plt.show()
```
At 100 kPa, the result is about 24.8 L, consistent with the approximate molar volume of an ideal gas near room temperature and atmospheric pressure.
````

```{admonition} Exercise 4.15
Use nested loops to simulate the kinetic model for every combination of `initial_concentrations = [0.50, 1.00]` mol/L and `rate_constants = [0.05, 0.10, 0.20]` s$^{-1}$. Use `dt = 0.01` s and `end_time = 10` s and print the final concentration of A for each combination.
```

````{admonition} Suggested solution
initial_concentrations = [0.50, 1.00]
rate_constants = [0.05, 0.10, 0.20]
dt = 0.01
end_time = 10
number_of_steps = int(end_time / dt)
for initial_concentration in initial_concentrations:
    for k in rate_constants:
        A = initial_concentration
        for _ in range(number_of_steps):
            change = k * A * dt
            A = A - change

```
````

```{admonition} Exercise 4.19

The exponential function can be approximated by $e^x\approx\sum_{i=0}^{n-1}x^i/i!$. Create `exp_taylor(x, n)` and use a loop to compare it with `math.exp(x)` for the integers 1–10.
```

````{admonition} Suggested solution
:class: tip, dropdown
```{code-block} Python
from math import factorial, exp

def exp_taylor(x, n):
    approximation = 0
    for i in range(n):
        approximation = approximation + x**i / factorial(i)
    return approximation

number_of_terms = 20
for x in range(1, 11):
    numerical = exp_taylor(x, number_of_terms)
    exact = exp(x)
    error = abs(numerical - exact)
    print(f"x = {x:2d}: approximation = {numerical:.6f}, error = {error:.2E}")
```
````
