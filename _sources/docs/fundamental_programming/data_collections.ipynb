Python provides several ways to organise data. The most important collections for our purposes are:

1. **lists** – flexible ordered collections that may contain similar or different kinds of data;
2. **arrays** – numerical collections that support efficient element-wise and vectorised calculations;
3. **tuples** – ordered collections that cannot be changed after they are created;
4. **dictionaries** – collections that associate keys with values.

We have already worked with lists. We now turn to arrays, tuples, and dictionaries.

## Arrays

NumPy arrays are particularly important in scientific programming. They may look similar to lists, but numerical operators behave differently.

`````python
import numpy as np

list1 = [1, 2, 3]
list2 = [2, 3, 1]

print("List addition:", list1 + list2)

array1 = np.array(list1)
array2 = np.array(list2)

print("Array addition:", array1 + array2)
`````

```{admonition} Exercise
:class: tip
Use the code above to explain the difference between list addition and array addition.
```

```{admonition} Suggested solution
:class: tip, dropdown
Adding two lists concatenates them: the second list is placed after the first. Adding two NumPy arrays of the same shape performs element-wise addition: `[1+2, 2+3, 3+1]`. This is analogous to vector addition.
```

## Tuples

A tuple is an ordered collection whose elements cannot be reassigned after the tuple has been created. This is useful for values that belong together and should remain fixed. One simple example is an RGB colour code, where red, green, and blue intensities are stored together. Tuples use parentheses rather than square brackets.

`````python
black = (0, 0, 0)
white = (255, 255, 255)
red = (255, 0, 0)
magenta = (255, 0, 255)

print(magenta)
`````

Tuples support indexing and iteration much like lists, but operations that would modify individual elements are not allowed. This makes the intention of fixed data explicit.

## Dictionaries

A dictionary stores **key–value pairs**. Unlike a list or array, which is usually accessed by numerical position, a dictionary is accessed by its keys. Keys are often strings, although other immutable types are also possible. Dictionaries use curly braces.

```{code-block} Python
dictionary = {"key1": value1, "key2": value2, "key3": value3}
```

Dictionaries are particularly useful for data in which values have natural labels: properties of elements, information about molecular species, experimental conditions, or metadata associated with samples.

`````python
atomic_masses = {"H": 1.01, "He": 4.00, "Li": 6.94, "Be": 9.01}

print(atomic_masses["Li"])  # atomic mass of lithium
`````
