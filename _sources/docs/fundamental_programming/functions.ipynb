## Definition

In addition to built-in Python functions such as `print` and `input`, we can define our own functions. This makes programs easier to organise and is particularly useful when we need to perform the same type of calculation several times.

Chemistry constantly uses relationships that take some quantities as input and produce a new quantity. From mass and molar mass we can find amount of substance; from hydronium-ion concentration we can find pH; and from pressure, volume, and temperature we can use the ideal-gas equation to calculate amount of substance. Such relationships are naturally represented as functions.

We begin with

$$n=\frac{m}{M},$$

where $n$ is the amount of substance in mol, $m$ is the mass in grams, and $M$ is the molar mass in g/mol.

`````python
def amount_in_moles(mass, molar_mass):
    amount = mass / molar_mass
    return amount
`````

The keyword `def` defines a function. Here the function is called `amount_in_moles`. The names in parentheses, `mass` and `molar_mass`, are *parameters*. Everything belonging to the function is indented.

`return` specifies the value that the function gives back. We can store that value in a variable or use it directly in another calculation. Merely defining a function produces no visible result; we must *call* it.

`````python
def amount_in_moles(mass, molar_mass):
    amount = mass / molar_mass
    return amount

moles_NaCl = amount_in_moles(5.85, 58.44)
print(f"The amount of NaCl is {moles_NaCl:.3f} mol.")

# We can also call the function directly inside print:
print(amount_in_moles(5.85, 58.44))
`````

The numbers `5.85` and `58.44` are *arguments*. When the function is called, these values are passed to the parameters `mass` and `molar_mass`. Parameters are therefore the names used in the function definition, whereas arguments are the actual values supplied in a function call.

## `return` and `print`

A function does not have to return a value. It can, for example, only print something:

`````python
def print_amount(mass, molar_mass):
    amount = mass / molar_mass
    print(f"The amount of substance is {amount:.3f} mol.")

print_amount(5.85, 58.44)
`````

Here we do not need `print` when calling the function because printing occurs inside it. Why, then, do we usually prefer `return` for calculated quantities?

If the result is needed later, the function must return a value. Printed output is text displayed to the user; it is not a numerical result that another part of the program can use. This program works because both function calls return numbers:

`````python
def amount_in_moles(mass, molar_mass):
    return mass / molar_mass

moles_NaCl = amount_in_moles(5.85, 58.44)
moles_KCl = amount_in_moles(7.46, 74.55)

total_amount = moles_NaCl + moles_KCl
print(total_amount)
`````

A function without an explicit `return` returns the special value `None`. As a general rule, it is therefore useful to return calculated values and decide separately how they should be displayed.

## Functions with several parameters

A function can have as many parameters as it needs. For an ideal gas,

$$P=\frac{nRT}{V}.


`````python
def ideal_gas_pressure(amount, temperature, volume):
    R = 8.314  # J/(mol K)
    pressure = amount * R * temperature / volume
    return pressure

# Function call 1: the arguments are numerical values
print(ideal_gas_pressure(0.10, 298.15, 0.0025))

# Function call 2: the arguments are variables
n = 0.10
T = 298.15
V = 0.0025
P = ideal_gas_pressure(n, T, V)
print(f"The pressure is {P/1000:.1f} kPa.")
`````

Variables outside the function do not need the same names as its parameters. With positional arguments, order determines where the values go: the first argument is assigned to `amount`, the second to `temperature`, and the third to `volume`. Clear scientific names nevertheless make code easier to read.

## Functions with several return values

Unlike a mathematical function in the elementary sense, a programming function can return several values. During a dilution, for example, we can calculate both the final concentration and the dilution factor.

`````python
def dilution(initial_concentration, initial_volume, final_volume):
    final_concentration = initial_concentration * initial_volume / final_volume
    dilution_factor = final_volume / initial_volume
    return final_concentration, dilution_factor

c2, factor = dilution(0.50, 10.0, 100.0)

print(f"The final concentration is {c2:.3f} mol/L.")
print(f"The solution has been diluted by a factor of {factor:.0f}.")
`````

The returned values are assigned to `c2` and `factor` in the same order in which they appear after `return`.

## Local and global variables

Variables defined inside a function are available only inside that function and are called *local variables*. Function parameters are local as well. Variables defined outside functions are *global variables*.

`````python
R = 8.314  # global variable

def ideal_gas_pressure(n, T, V):
    pressure = n * R * T / V  # local variable
    return pressure

P = ideal_gas_pressure(0.10, 298.15, 0.0025)
print(P)
`````

Here `R` is global and can be read both outside and inside the function. The parameters `n`, `T`, and `V`, and the variable `pressure`, are local. `P` is global because it is created outside the function. What happens if we try to use the local variable `pressure` outside the function?

`````python
R = 8.314

def ideal_gas_pressure(n, T, V):
    pressure = n * R * T / V
    return pressure

P = ideal_gas_pressure(0.10, 298.15, 0.0025)
print(pressure)  # raises NameError
`````

We obtain an error even though `pressure` is visibly defined in the function. The variable exists only locally while the function is executing. Its value is nevertheless available outside because the function returns it and we store it in `P`.

Python has a `global` keyword that can be used to modify a global variable inside a function. It is rarely necessary and often makes changes to program state harder to follow. Passing values in as arguments and returning results is usually clearer, easier to test, and easier to reuse.

## Default parameter values

Sometimes a parameter usually has one value but should remain adjustable. We can then give it a default. In measured data it is rarely sensible to demand that two floating-point values be exactly equal, so we might ask whether a measured pH lies within a chosen tolerance around 7.

`````python
def is_neutral(pH, tolerance=0.1):
    return 7 - tolerance <= pH <= 7 + tolerance

print(is_neutral(7.05))       # uses the default value 0.1
print(is_neutral(7.15, 0.2))  # uses a tolerance of 0.2
`````

Parameters with default values are normally placed after required parameters. The function can be called either with or without a separate argument for `tolerance`.
