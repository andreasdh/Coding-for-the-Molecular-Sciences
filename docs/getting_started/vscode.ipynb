# Visual Studio Code for molecular scientists

**Visual Studio Code**, usually called **VS Code**, is a popular and powerful editor that can be used for both ordinary Python programs and Jupyter notebooks. VS Code is a large application, but at first you only need a small part of its functionality. The aim of this chapter is therefore not to turn you into a software developer. The aim is to help you open, run, modify, and organise Python code safely and confidently so that you can concentrate on the science.

```{admonition} Learning outcomes
After working through this chapter, you should be able to:

1. open a working folder and recognise the most important parts of VS Code
2. select the correct Python environment and the correct kernel in a Jupyter notebook
3. create, save, and run an ordinary Python file (`.py`)
4. create and work with a Jupyter notebook (`.ipynb`)
5. run complete programs and individual cells in a controlled way
6. install and import common Python libraries
7. organise code and data so that relative file paths work
8. use error messages and a few simple strategies to locate errors
```

```{admonition} What you do not need to learn yet
:class: note
VS Code contains tools for version control, automated testing, advanced debugging, remote execution, and development of large software projects. These may be useful later, but they are not necessary for getting started with scientific programming.

For now, you mainly need the **Explorer**, **Editor**, **Terminal**, selection of a **Python interpreter**, and selection of a **kernel** in notebooks.
```

## What is VS Code?

VS Code is an editor that can be extended with support for different programming languages and file types. When we install the Python and Jupyter extensions, we can use VS Code to:

- write and run Python files
- open and run Jupyter notebooks
- read error messages in an integrated terminal
- inspect variables, tables, and graphs
- receive suggestions for function names and documentation while we type

VS Code is not the same as **Visual Studio**, even though the names are similar. VS Code is the application we use here.

```{admonition} The editor and Python
**VS Code** is the workspace in which you write and organise code. **Python** is the language whose interpreter reads and executes the instructions. You therefore need both VS Code and a Python installation.
```

## 4. Python environments, interpreters, and kernels

This is the part of the setup that causes the most problems for new users. The concepts are simpler than they may first appear.

```{admonition} Python interpreter
An **interpreter** is the Python program that reads and runs your code. A computer can have several Python interpreters installed at the same time.
```

```{admonition} Python environment
A **Python environment** consists of a Python interpreter and the libraries installed with it. One environment may, for example, contain NumPy and Pandas while another does not.
```

```{admonition} Kernel
A **kernel** is the process that runs code cells in a Jupyter notebook. For a Python notebook, we usually select one of our Python environments as the kernel.
```

In a `.py` file, you therefore need to select the correct **interpreter**. In an `.ipynb` file, you need to select the correct **kernel**. In practice, both should normally point to the same Python environment for your course or project.

```{admonition} A common problem
:class: warning
If `import numpy` works in one file but fails in another, the files are often using different Python environments. Check the selected interpreter or kernel before reinstalling the library.
```

### Create an environment for your work

If you have not been given a ready-made environment or installation recipe, you can create a local environment in the working folder:

1. Open the Command Palette.
2. Select **Python: Create Environment**.
3. Select **Venv**.
4. Select the Python version you have installed.
5. Wait for VS Code to create the `.venv` folder.
6. Check that this environment is selected in the Status Bar. If not, use **Python: Select Interpreter** and choose the environment containing `.venv`.

A `.venv` folder is an isolated environment for this particular working folder. It reduces the risk that libraries from different courses and projects will conflict.

```{admonition} If you use Anaconda or Miniconda
:class: note
If you have been given an environment through Anaconda (a Conda environment), you should normally select it using **Python: Select Interpreter** rather than creating an additional `.venv`. You do not need to use both systems at the same time.
```

### Install libraries in the correct environment

Select **Terminal → New Terminal**. When the correct environment is selected, VS Code will usually activate it automatically in a new terminal. The environment name, for example `(.venv)`, is often shown at the beginning of the command line.

Libraries that we use frequently can be installed with:

```bash
python -m pip install numpy matplotlib pandas scipy seaborn sympy
```

The command `python -m pip` asks the selected Python interpreter to use its own package installer. This makes it less likely that libraries will be installed in a different environment from the one you are actually using.

```{admonition} Do not reinstall immediately
:class: tip
If an import fails, first check the interpreter or kernel. `ModuleNotFoundError` often means that the correct library exists in another environment, not that it is entirely absent from the computer.
```

## 5. Create and run a Python file

We will begin with an ordinary Python program.

1. Click **New File** in Explorer.
2. Name the file `first_program.py`. The `.py` extension tells VS Code that the file contains Python code.
3. Type or paste in the program below.
4. Save the file.

The program calculates the amount of NaCl from its mass and molar mass.

`````python
mass_g = 2.922
molar_mass_g_mol = 58.44
amount_mol = mass_g / molar_mass_g_mol

print(f"Amount of NaCl: {amount_mol:.3f} mol")
`````

### Run the complete program

With `first_program.py` open, click the triangular **Run Python File in Terminal** button in the upper-right corner of the editor. VS Code then saves and runs the file with the selected interpreter. The result appears in the Terminal at the bottom.

You can also right-click in the editor and select **Run Python File in Terminal**. When you are working with a continuous program, you should periodically run the entire file from the beginning. This reveals whether the program depends on variables that exist only because you previously ran individual lines.

### Run parts of the program

Select one or more lines and press `Shift+Enter`. The code is sent to an interactive Python terminal. This is useful when you want to try a small calculation without running the entire program.

The interactive terminal remembers variables you have already created. This is convenient, but it can also hide errors. If the program works only after you have run lines in a particular, accidental order, run the entire file again.

### Cells in a `.py` file

If you like working with smaller sections of a program, you can mark cells in an ordinary Python file using `# %%`. VS Code will then display a **Run Cell** button above each section. This can be a useful transition between ordinary Python files and notebooks.

`````python
# %%
temperatures_C = [20, 30, 40, 50]
temperatures_K = [temperature + 273.15 for temperature in temperatures_C]

# %%
print(temperatures_K)
`````

### A chemical example in a notebook

The program below uses a simple first-order model to calculate how the concentration of a reactant decreases with time:

$$c(t)=c_0e^{-kt}$$

Run the cell with `Shift+Enter`. The figure appears directly below the cell.

`````python
import numpy as np
import matplotlib.pyplot as plt

time_s = np.linspace(0, 120, 100)
initial_concentration_M = 0.100
rate_constant_s = 0.025
concentration_M = initial_concentration_M * np.exp(-rate_constant_s * time_s)

plt.plot(time_s, concentration_M)
plt.xlabel("Time (s)")
plt.ylabel("Concentration (mol/L)")
plt.title("First-order reaction")
plt.grid(alpha=0.3)
plt.show()
`````

### The kernel remembers what you have run

When you run a code cell, the kernel stores the variables in memory. A later cell can therefore use variables from an earlier cell. This makes it easy to build an analysis step by step.

But cells can also be run in a different order from the one in which they appear. A notebook can therefore display a result that cannot be reproduced when someone else opens it. A good notebook should be able to run from top to bottom.

Use this routine regularly:

1. Save the notebook.
2. Restart the kernel with **Restart**.
3. Run all cells from the top with **Run All**.
4. Check the first error if execution stops.

```{admonition} Reproducible notebook
A reproducible notebook can be run from the first to the last cell in the correct order and produce the expected results without hidden intermediate steps.
```

## 8. Folders, data files, and file paths

When we begin working with data, the program needs to know where the data file is located. The safest workflow is to keep the notebook and data in a clear project folder and use **relative file paths**.

If the notebook is in `data_handling`, and the data file is in the subfolder `data`, the file can be read like this:

```python
import pandas as pd

measurements = pd.read_csv("data/uvvis.csv")
```

This path means 'enter the `data` folder from the location where the program is working, and open `uvvis.csv`'. Do not write a long path that exists only on your own computer, for example one beginning with your user name. Such code will often stop working when the folder is moved or shared.

The cell below shows which folder Python is using as its current working directory.

`````python
from pathlib import Path

working_directory = Path.cwd()
print("Working directory:", working_directory)
`````

```{admonition} If Python cannot find the data file
:class: tip
Check these three things before changing the code substantially:

1. Did you open the working folder with **File → Open Folder**?
2. Is the data file actually located in the folder you specified?
3. Are the filename, capitalisation, and file extension written correctly?

`FileNotFoundError` usually means that Python is looking in the wrong place or for a slightly different filename.
```

## 9. When something does not work

Error messages are a normal part of programming. First read the bottom of the error message. This usually states the error type and gives a short description. Then look for the line in your own code to which the error points.

| Error or problem | Common explanation | What should you check? |
|---|---|---|
| `NameError` | A variable name does not exist | Spelling, capitalisation, or a cell that has not been run |
| `SyntaxError` | Python does not understand the syntax | Missing parenthesis, colon, quotation mark, or incorrect indentation |
| `ModuleNotFoundError` | The library is not in the active environment | Check the interpreter/kernel before installing |
| `FileNotFoundError` | Python cannot find the data file | Working folder, relative path, and exact filename |
| A notebook cell will not run | The kernel is missing or has stopped | Select a kernel or use **Restart** |
| Red or yellow lines under an import | VS Code is examining a different environment | Select the correct interpreter and, if necessary, reload the window |
| The program produces a number, but it seems wrong | Code can be valid even when the model is wrong | Units, input data, signs, model, and reasonableness checks |

### A simple debugging strategy

1. Find the first error, not the last consequence.
2. Read the final line of the error message.
3. Check the line it points to.
4. Print intermediate values with `print`, or display variables in the notebook.
5. Test a smaller part of the code.
6. Restart the interpreter or kernel and run from the top.
7. Ask for help and show the code, the complete error message, and what you expected to happen.

```{admonition} Code that runs can still be chemically wrong
:class: warning
VS Code and Python can detect invalid syntax and many technical problems. They will not normally detect that you mixed grams and kilograms, used Celsius where a model requires kelvin, assigned the wrong unit to a concentration, or selected a model outside its range of validity.

When the program runs, the scientific checking begins: Are the units consistent? Is the order of magnitude reasonable? Does the model behave sensibly when you change a parameter?
```

## 11. Good working habits

A few habits make programming easier to manage and the results easier to check:

- Open the project folder, not just an individual file.
- Use descriptive filenames such as `titration_analysis.ipynb` and `reaction_rate.py`.
- Prefer lower-case letters and underscores in filenames.
- Keep original data separate from code and figures, for example in `data` and `figures` folders.
- Do not overwrite raw data when you clean a data set.
- Include units in variable names or comments when useful.
- Save often.
- Run the entire `.py` file or use **Restart → Run All** before submitting or sharing a notebook.
- Read through the code, figures, and scientific explanations as if a colleague were going to reproduce the work.

````{admonition} Suggested folder structure for an analysis
```text
experiment/
├── analysis.ipynb
├── data/
│   └── raw_measurements.csv
└── figures/
    └── calibration_curve.png
```
````

## When should you use `.py`, and when should you use `.ipynb`?

There is no single file type that is always best. Choose according to what you are doing.

| Prefer `.py` when ... | Prefer `.ipynb` when ... |
|---|---|
| the program should run as a whole from start to finish | you are exploring data step by step |
| you are practising variables, conditionals, loops, and functions | you want to combine code, figures, and scientific explanations |
| an analysis or simulation will be repeated for many files | you are documenting the choices in one analysis |
| you want to reuse the program as a tool | you are trying models and comparing results |

In practice, the two can also be combined. A notebook can import functions from a `.py` file, and a `.py` file can be divided into interactive cells using `# %%`. The important point is that the workflow is clear and reproducible.
