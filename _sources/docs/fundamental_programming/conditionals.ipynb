## Definition

```{admonition} Conditional
A conditional is a logical test used to determine whether a criterion is satisfied. If the criterion is satisfied, an action is performed. If it is not satisfied, that action is not performed. In Python, such a decision normally begins with `if`.
```

Conditionals are central to programming and to scientific reasoning. We often draw conclusions that depend on one or more measured or calculated quantities: if the pH is below a threshold, we classify a solution in one way; if an absorbance lies outside a calibration range, the sample may need to be diluted or measured again; if a reaction quotient is smaller than the equilibrium constant, the system will move in one direction towards equilibrium.

A program can apply such decisions consistently, but we must decide which criteria are scientifically meaningful. A conditional is therefore also a way of making our scientific assumptions explicit.

For a simplified example near 25 °C, we may call a solution acidic if its pH is below 7. In pseudocode:

```{code-block} text
if pH is less than 7:
    classification = "acidic"
otherwise:
    classification = "not acidic"
```

The Python structure is very similar:

`````python
if pH < 7:
    classification = "acidic"
else:
    classification = "not acidic"
`````

## Conditionals in Python

Let us return to the pH example and make the pH user input:

`````python
pH = float(input("Enter the pH value: "))

if pH < 7:
    print("The solution is acidic.")
`````

The program asks the user for a pH value and converts it to a float. A conditional begins with `if`, followed by a logical test. Here we test whether the pH is less than 7. If the test is true, the message is printed. A colon follows the test, and everything belonging to the conditional is indented. If the test is false, nothing happens.

The test itself always has one of the Boolean values `True` or `False`:

`````python
pH = 5.5
print(pH < 7)   # Prints True
print(pH > 7)   # Prints False
`````

If we want something to happen when the test is false, we can add `else`:

`````python
pH = float(input("Enter the pH value: "))

if pH < 7:
    print("The solution is acidic.")
else:
    print("The solution is not acidic.")
`````

We can add further criteria with `elif`, short for *else if*:

`````python
pH = float(input("Enter the pH value: "))

if pH < 7:
    print("The solution is acidic.")
elif pH > 7:
    print("The solution is basic.")
else:
    print("The solution is neutral.")
`````

If pH is neither below nor above 7, it follows logically that it is 7, so `else` needs no further test. If we do want to test equality explicitly, we use `==`:

`````python
if pH == 7:
    print("The pH value is exactly 7.")
`````

`==` tests whether two values are equal. A single equals sign (`=`) assigns a value to a variable. The comparison operators are summarised below.

| Symbol | Meaning |
|---|---|
| `>` | Greater than |
| `<` | Less than |
| `==` | Equal to |
| `<=` | Less than or equal to |
| `>=` | Greater than or equal to |
| `!=` | Not equal to |

Some important points:

- Indented statements are executed only when the test above them selects that branch. Indentation is part of Python's structure, not decoration.
- `elif` tests another criterion, while `else` is used if none of the preceding criteria are true.
- Only the first true branch in one continuous `if`–`elif`–`else` structure is executed. The order of the tests can therefore affect the result.
- A conditional must begin with `if`; `elif` and `else` are optional.
- A new `if` begins a new, independent test.

### When numbers are almost equal

Measured and calculated floating-point values are rarely exact. A pH that should theoretically be 7 might, for example, be stored as `6.999999999`. The test `pH == 7` would then be false. It is often better to test whether the value lies within a reasonable tolerance:

`````python
pH = 6.99
tolerance = 0.01

if abs(pH - 7) <= tolerance:
    print("The solution is approximately neutral.")
`````

The appropriate tolerance depends on measurement uncertainty and the purpose of the calculation. In addition, the pH of a neutral solution depends on temperature. Treating pH 7 as neutral is therefore an approximation appropriate near 25 °C.

## Compound conditions

Sometimes several criteria must be considered at once. We can combine tests using `and`, `or`, and `not`.

| Operator | Meaning |
|---|---|
| `and` | Both criteria must be true |
| `or` | At least one criterion must be true |
| `not` | Reverses a Boolean value |

Suppose an analytical method has a recommended absorbance range from 0.10 to 1.00. We can test whether a measurement lies in the range like this:

`````python
absorbance = 0.82

if absorbance >= 0.10 and absorbance <= 1.00:
    print("The measurement is within the recommended range.")
else:
    print("Consider adjusting the sample or measurement conditions.")
`````

Python also lets us write the same test as a chained comparison:

`````python
if 0.10 <= absorbance <= 1.00:
    print("The measurement is within the recommended range.")
`````

The numerical limits are only an example. The valid range must be determined from the actual instrument, calibration, and analytical method. With `or`, we can test whether the absorbance lies outside the range:

`````python
if absorbance < 0.10 or absorbance > 1.00:
    print("The measurement is outside the recommended range.")
`````

## Nested conditionals

A conditional can contain another conditional. This is useful when a second decision is relevant only if an earlier criterion has been satisfied. Each new level receives additional indentation.

The example below performs a simplified quality check of a spectrophotometric measurement. The absorbance is assessed only if the measurement has already been blank-corrected.

`````python
blank_corrected = input("Has the measurement been blank-corrected (yes/no)? " ).strip().lower()

if blank_corrected == "yes":
    absorbance = float(input("Enter the absorbance: "))

    if 0.10 <= absorbance <= 1.00:
        print("The measurement is within the recommended range.")
    else:
        print("Consider diluting the sample or changing the measurement conditions.")
elif blank_corrected == "no":
    print("Blank-correct the measurement before assessing the absorbance.")
else:
    print("Please answer 'yes' or 'no'.")
`````

The inner `if`–`else` structure belongs to the first branch and is therefore considered only when the user answers `yes`. The methods `strip()` and `lower()` remove surrounding whitespace and convert the text to lower case. Thus, for example, ` Yes ` is interpreted as `yes`.

Nested structures are useful when tests genuinely depend on one another, but too many levels make code difficult to read. In that case, it may be better to divide the problem into several steps or use compound conditions.
