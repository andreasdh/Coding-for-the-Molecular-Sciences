# Introduction: Programming as a Tool in Chemistry

Chemistry is an experimental science. We observe, measure, compare, and try to explain why substances behave as they do. Traditionally, the laboratory has been the main arena for this work. There we can mix substances, change the temperature, measure pH, or investigate how quickly a reaction proceeds.

But not all chemical phenomena can be observed directly. We cannot see how electrons are distributed in a molecule, follow every individual collision between molecules, or wait a few million years for a slow geochemical process to finish. Experiments can also produce such large amounts of data that we need help finding patterns in them. In these situations, the computer can serve as a laboratory, a visualisation tool, and a computational aid.

Programming gives us a language for telling the computer what to do. In chemistry, however, the main goal is not to create computer programs. The goal is to use programs to ask better questions, investigate chemical relationships, and understand what models and experimental data are telling us.

## Why programming?

A computer is very good at performing simple operations quickly and repeatedly. It can process thousands of measurements, repeat the same calculation for many different temperatures, or calculate how the concentrations of several substances change in small time steps. This is useful because many chemical problems consist precisely of many simple processes acting together.

Consider, for example, a spectrometer that measures how much light a solution absorbs at several hundred wavelengths. A single measurement is not very difficult to handle, but if we have measurements from many solutions, the amount of information quickly becomes substantial. With a program, we can read all the data, remove or deal with missing values, visualise the spectra, and compare them.

Or consider a chemical reaction. A rate law describes how quickly the concentration is changing at this moment. If we want to find the concentration one minute from now, we need to investigate how all the small changes accumulate over time. The computer can divide the reaction into many small time steps and update the concentrations at each step.

The computer does not, however, make chemical judgements on its own. It does not know whether the model is reasonable, whether the units are correct, or whether the result makes chemical sense. We have to decide that.

## A tool for thinking

Programming is often used to make calculations that we already know how to perform more efficient. That is useful, but programming can also influence how we think about a problem.

In mathematics, we often describe a relationship with a finished function or equation. Program code, by contrast, must contain a recipe for what should happen, step by step. We therefore have to make our reasoning explicit. If, for example, we want to simulate a chemical reaction, a very simple algorithm might look like this:

```text
set the initial concentrations

repeat many times:
    calculate the reaction rate
    calculate how much the concentrations change
    update the concentrations
```

Here we see the chemical process in a different way than in a finished mathematical expression. The new concentration is the old concentration plus a small change. The new concentration is then used to calculate what happens in the next time step. In this way, the chemical system develops in the program in roughly the same way as we imagine the reaction developing over time.

This does not mean that code always gives better understanding than mathematics. Mathematical expressions, program code, graphs, and molecular models show different aspects of the same phenomenon. The strength lies in being able to move between them and see how they are connected.

## From chemical phenomenon to program

Before we can simulate a chemical system, we need to describe the system with a model. A model is a simplified representation of reality. It includes some properties and leaves out others.

When we create a program to explore chemistry, we often go through the following process:

1. **We formulate a question.** How does the reaction rate change with temperature? Why does a protein fold in a particular way?
2. **We define the system.** Which particles, compounds, or variables are relevant?
3. **We choose or create a model.** The model might, for example, be a rate law, an approximation of the forces between atoms, or a statistical model for experimental data.
4. **We translate the model into an algorithm and program code.** The assumptions and operations must then be expressed in a form the computer can follow.
5. **We examine the result.** We may obtain numbers, tables, graphs, molecular structures, or an animation of the system.
6. **We evaluate and improve the model.** Does the result agree with chemical theory and experimental data? What happens if we change a parameter?

The process therefore does not stop when the program runs without error messages. A program can work perfectly well while still being based on a poor model. The output of the program must therefore be used as a starting point for further chemical thinking.

```{admonition} Model
A model is a simplified representation of a phenomenon or system. A model is based on assumptions and therefore always has a particular range of validity and certain limitations.
```

## Simulations – experiments on a computer

A simulation is a program that calculates how a model behaves under specified conditions. We can, for example, simulate how molecules move, how a drug binds to a protein, or how the concentration of ozone in the atmosphere changes over time.

Simulations make it possible to perform numerical experiments. We can change the temperature, the initial concentration, or the strength of an interaction and investigate what happens. If the result is surprising, we can modify the model or examine the system more closely. The computer often gives us rapid feedback, and this movement between questions, results, and new questions can drive the investigation forward.

It is also possible to investigate systems that are too small, large, fast, slow, expensive, or dangerous for an ordinary laboratory. A simulation can show us a microscopic level that we cannot otherwise observe directly, such as the motion of molecules or the distribution of electrons in a chemical compound.

A simulation is not, however, a direct window onto reality. It shows how the **model** behaves. If important interactions are missing from the model, they will also be missing from the simulation. Simulations should therefore be compared with chemical theory, experimental results, and other models. They do not necessarily replace laboratory experiments, but they can complement them by making processes visible, suggesting explanations, and providing a basis for new experiments.

```{admonition} Simulation
A simulation is a calculation of how a model behaves. The simulation can be used as a numerical experiment in which we change conditions and investigate the consequences.
```

## From measurement to knowledge – data handling

Experimental data do not explain themselves. Before we can draw a conclusion, we often need to organise, clean, visualise, and analyse the measurements. This may involve:

- reading data from instruments and data files
- identifying missing or unreasonable values
- calculating means, standard deviations, and other statistical quantities
- visualising data with graphs
- fitting models to measured data points
- comparing results from several experiments

Programming makes it possible to carry out the same operations systematically and reproducibly. If the analysis is written as code, we can see which choices were made and run the same analysis again. This is particularly important when the data set is large or the analysis consists of many steps.

Data handling is not merely about finding a function that fits the data points. We must also ask what the variables represent, how uncertain the measurements are, and whether the model has a chemical justification. An excellent mathematical fit is not necessarily a good chemical explanation.

```{admonition} Data handling
Data handling means organising, processing, visualising, and analysing data so that we can investigate patterns and draw well-founded conclusions.
```

## Computers require explicit assumptions

When we calculate by hand, we can skip intermediate steps because we understand what we mean. A computer, by contrast, needs precise instructions. Every variable must have a value, every operation must occur in the correct order, and every special case must be handled.

This can feel cumbersome, but it is also a strength. When we translate a chemical problem into code, we are forced to state what the quantities mean and how they are related. We often discover ambiguities in our own understanding when we try to explain the problem to the computer.

Error messages and unexpected results are therefore not merely obstacles. They can be clues. Perhaps we have misspelled a variable name, but perhaps we have also forgotten an important chemical process or used a model outside its range of validity. Debugging a program means investigating both the code and the scientific reasoning behind it.

## AI can write code – but who does the scientific thinking?

Generative artificial intelligence (AI) can now create, explain, and correct program code. This lowers the threshold for using programming in chemistry. A chemist does not necessarily need to remember all the syntax or build every program from scratch. AI can, for example, help us read a data file, create a plot, or locate an error in a loop. This can leave us more time for the chemical problem we actually want to investigate.

This does not mean that understanding programming has become unnecessary. AI can generate code that looks convincing but uses incorrect units, an unsuitable model, or a function that does not do what the AI claims. It can also propose a very complicated solution to a simple problem. If we do not understand what the program is doing, it becomes difficult to decide whether the result is correct.

The most productive use of AI therefore occurs when we retain control of the problem-solving process ourselves. We should formulate the chemical question, define the system, and evaluate which assumptions are reasonable. AI can then act as a dynamic programming assistant that offers suggestions, explains code, and helps us move forward when we get stuck.

```{admonition} Think first, ask second – and then think again
When you use AI as a programming assistant, you can work like this:

1. First explain to yourself what the problem is and what the program should do.
2. Give the AI relevant chemical context, units, and clear constraints.
3. Consider asking for small parts of the program or an explanation rather than necessarily requesting a complete solution.
4. Read and test the code. Change values and investigate whether the result behaves as expected.
5. Evaluate the result using your chemical knowledge. You, not the AI, are responsible for the scientific conclusion.
```

## What does a chemist need to know about programming?

You do not need to become a software developer to benefit from programming in chemistry. But you should be able to:

- read simple code and explain what the most important lines do
- break a problem into smaller steps
- recognise variables, conditionals, loops, functions, and data collections
- run and modify existing programs
- test whether the code produces reasonable results
- interpret graphs, numbers, and other results in a chemical context
- evaluate the assumptions and limitations of a model
- use documentation, fellow students or colleagues, and AI for help without giving up control of the reasoning

The goal is therefore not to memorise as many commands as possible. The goal is to be able to use program code as part of chemical problem solving. The rest of these pages will give you the tools you need to get started.

## A first program – from concentration to pH

Let us begin with a small program that calculates pH from the concentration of hydronium ions, $\mathrm{H_3O^+}$. The relationship is:

$$
\mathrm{pH}=-\log_{10}\left[\mathrm{H_3O^+}\right]
$$

Here we use the common approximation in which pH is calculated directly from concentration. Strictly speaking, pH is defined in terms of hydronium-ion activity. The difference becomes more important in more concentrated solutions, but we will ignore it in this first example.

Before you run the program below, try reading it as a recipe:

1. The program imports a mathematical function for calculating logarithms.
2. The concentration of hydronium ions is stored in a variable.
3. The program uses the model above to calculate pH.
4. The result is printed.

````python
from math import log10

h3o_concentration = 0.001  # mol/L
pH = -log10(h3o_concentration)

print("pH =", pH)
````

The program gives pH = 3.0. But the interesting point is not only that the program finds the answer. We can use the program to investigate the relationship between concentration and pH.

## Summary

Programming can be used to automate calculations, but its most important role in chemistry is broader than that. With code, we can make models explicit, simulate systems that we cannot observe directly, and find patterns in large amounts of experimental data. Programming also gives us new representations of chemical phenomena and can therefore help us see connections between mathematics, the microscopic level, and what we observe in the laboratory.

AI makes it easier to create and understand code, including for scientists whose main interest is chemistry. But scientific responsibility cannot be handed over to either the computer or the AI. We still have to ask the questions, choose and evaluate the models, and decide whether the results make chemical sense.
