# Programming tools

The core programming language is only one part of a scientific workflow. We also need tools for combining code with explanations, keeping track of files, and preserving the history of our work. Two particularly useful tools are **Jupyter notebooks** and **Git/GitHub**.

The practical VS Code chapter earlier in this book already showed how to create and run Jupyter notebooks. Here we focus briefly on what these tools contribute to scientific work.

## Jupyter notebooks

A Jupyter notebook combines Markdown text, executable code, figures, tables, and output in one document. This makes notebooks useful for exploratory data analysis and for documenting how a scientific result was obtained.

A well-structured scientific notebook should normally:

- explain the scientific question and assumptions in Markdown cells;
- keep units and variable meanings clear;
- separate data loading, processing, analysis, and visualisation into understandable steps;
- run reproducibly from the first cell to the last;
- distinguish interpretation from the code that produces a result.

```{admonition} A useful reproducibility check
Before sharing a notebook, restart the kernel and run all cells from the top. If the notebook fails, it probably depends on hidden state from an earlier interactive session.
```

## Git and GitHub

**Git** is a version-control system. It records snapshots of files so that changes can be compared, combined, and reversed. **GitHub** is a platform for storing Git repositories online and collaborating around them.

Version control is more deliberate than ordinary cloud-file synchronisation. Instead of merely replacing a file with its newest version, Git stores a history of meaningful changes. This is particularly valuable when code, data-processing scripts, figures, and documentation develop together.

For scientific work, version control can help you:

- recover an earlier working version of an analysis;
- see exactly what changed between two versions;
- collaborate without repeatedly exchanging files such as `analysis_final_v2_really_final.ipynb`;
- document the development of code used to produce scientific results;
- publish reusable scripts, notebooks, and teaching material.

You do not need to master advanced Git workflows before using programming productively in molecular science. The important idea is that scientific code is worth treating as a developing, documented object rather than as a disposable calculation.

## Choosing the right level of tooling

For a small exercise, a single `.py` file or notebook may be enough. For a larger analysis, a clear folder structure and version control become increasingly valuable. As projects grow, functions, reusable modules, tests, and collaborative Git workflows can be introduced gradually.

The goal is not to imitate professional software development for its own sake. The goal is to use enough structure that your scientific work remains understandable, reproducible, and easy to improve.
